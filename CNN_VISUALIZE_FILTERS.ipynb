```python
# Example 4.17 CNN Visualize Filters.ipynb
#Modified from:

#https://machinelearningmastery.com/how-to-visualize-filters-and-feature-maps-in-convolutional-neural-networks/
# load vgg model
from keras.applications.vgg19 import VGG19
# load the model
model = VGG19()
# summarize the model
model.summary()
```

```python
n=0
for layer in model.layers:
  print(n,layer.name)
  n+=1
```

```python
n=0
for layer in model.layers:
  if 'conv' in layer.name:
    filters, biases = layer.get_weights()
    print(n,layer.name, filters.shape, biases.shape)
  n+=1
```

```python
from matplotlib import pyplot
# retrieve weights from the first hidden layer
n = 1
filters, biases = model.layers[n].get_weights()
s = filters.shape
print("Color channels: ", s[0])
print("Filter size: ", s[1],s[2])

print("Total number of filters : ", s[3])
# normalize filter values to 0-1 so we can visualize them
f_min, f_max = filters.min(), filters.max()
filters = (filters - f_min) / (f_max - f_min)
# plot first few filters
#n_filters, ix = s[3], 1
n_filters, ix = 4, 1

pyplot.figure(figsize=(10,10))
for i in range(n_filters):
  # get the filter
  f = filters[:, :, :, i]
  # plot each channel separately
  for j in range(s[0]):
    ax = pyplot.subplot(n_filters, s[0], ix)
    ax.set_xticks([])
    ax.set_yticks([])
    pyplot.imshow(f[j,:, :], cmap='gray')
    ix += 1
# show the figure
pyplot.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

from tensorflow.keras.applications.vgg19 import VGG19, preprocess_input
from tensorflow.keras.utils import load_img, img_to_array
from tensorflow.keras.models import Model
from google.colab import files

# Upload gambar dari komputer
uploaded = files.upload()

# Mengambil nama file gambar yang di-upload
image_path = next(iter(uploaded))

# Fungsi untuk menampilkan feature maps
def plot_feature_maps(feature_maps):
    col = 8
    total_maps = feature_maps.shape[3]
    row = int(np.ceil(total_maps / col))

    ix = 1
    plt.figure(figsize=(20, 2.5 * row))

    for i in range(total_maps):
        ax = plt.subplot(row, col, ix)
        ax.set_xticks([])
        ax.set_yticks([])

        plt.imshow(feature_maps[0, :, :, i], cmap='gray')
        ix += 1

    plt.tight_layout()
    plt.show()

# Memuat model VGG19
model = VGG19(weights='imagenet')

# Memilih hidden layer yang ingin divisualisasikan
n = 1

# Membuat model baru untuk mengambil output layer terpilih
model = Model(
    inputs=model.inputs,
    outputs=model.layers[n].output
)

# Menampilkan ringkasan model
model.summary()

# Memuat gambar dengan ukuran 224 x 224
img = load_img(image_path, target_size=(224, 224))

# Mengubah gambar menjadi array
img = img_to_array(img)

# Menambahkan dimensi batch
img = np.expand_dims(img, axis=0)

# Preprocessing gambar untuk VGG19
img = preprocess_input(img)

# Mengambil feature maps
feature_maps = model.predict(img)

print("Feature maps:", feature_maps.shape)

# Menampilkan semua feature maps
plot_feature_maps(feature_maps)
```

```python

import matplotlib.pyplot as plt
import numpy as np

from tensorflow.keras.applications.vgg19 import VGG19, preprocess_input
from tensorflow.keras.utils import load_img, img_to_array
from tensorflow.keras.models import Model
from google.colab import files

# Upload gambar
uploaded = files.upload()
image_path = next(iter(uploaded))

# Memuat model VGG19
base_model = VGG19(weights='imagenet')

# Memilih layer konvolusi
n = 2  # Ubah menjadi 4 atau 5 untuk layer konvolusi berikutnya

# Membuat model untuk mengambil output layer terpilih
model = Model(
    inputs=base_model.inputs,
    outputs=base_model.layers[n].output
)

# Menampilkan nama layer
print("Layer yang dipilih:", base_model.layers[n].name)
model.summary()

# Memuat dan mengubah ukuran gambar
img = load_img(image_path, target_size=(224, 224))
img = img_to_array(img)
img = np.expand_dims(img, axis=0)

# Preprocessing gambar
img = preprocess_input(img)

# Menghasilkan feature maps
feature_maps = model.predict(img)
print("Feature maps:", feature_maps.shape)

# Menampilkan feature maps
def plot_feature_maps(feature_maps):
    total_maps = feature_maps.shape[3]
    col = 8
    row = int(np.ceil(total_maps / col))

    plt.figure(figsize=(16, 2 * row))

    for i in range(total_maps):
        ax = plt.subplot(row, col, i + 1)
        ax.set_xticks([])
        ax.set_yticks([])
        plt.imshow(feature_maps[0, :, :, i], cmap='gray')

    plt.suptitle("Feature Maps - " + base_model.layers[n].name)
    plt.tight_layout()
    plt.show()

plot_feature_maps(feature_maps)
```
